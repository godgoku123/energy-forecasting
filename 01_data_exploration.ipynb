```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data/demanddata_2025.csv")

df.head()
```

```python
df.info()
```

```python
df.columns
```

```python
# Select the first 336 half-hourly observations
week = df["ND"].iloc[:336]

plt.figure(figsize=(12, 5))

plt.plot(week.to_numpy())

plt.title("Electricity Demand Over One Week")
plt.xlabel("Half-hourly period")
plt.ylabel("Electricity Demand (MW)")

plt.grid(True)
plt.show()
```
