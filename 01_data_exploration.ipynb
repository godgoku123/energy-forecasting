Research question 1: Does electricity demand differ between weekdays and weekends?
Hypothesis: I predict electricity demand to rise earlier on weekdays as people generally begin school and work earlier than on weekends. I also expect overall weekday demand to be higher because of commercial and industrial activity.

```python
#Comparing weekday and weekend electricity demand
df["SETTLEMENT_DATE"] = pd.to_datetime(
    df["SETTLEMENT_DATE"],
    format="mixed",
    dayfirst=True
)
```

```python
df["DAY_OF_WEEK"] = df["SETTLEMENT_DATE"].dt.dayofweek #extracting day of week as a number i.e monday = 0 ... sunday = 6.
df["DAY_TYPE"] = df["DAY_OF_WEEK"].apply(
    lambda day: "Weekday" if day < 5 else "Weekend"
)
```

```python
comparison = df.groupby(
    ["SETTLEMENT_PERIOD", "DAY_TYPE"]
)["ND"].mean() #getting mean ND for "Weekday" and for "Weekend"

comparison = comparison.unstack("DAY_TYPE")
```

```python
comparison = comparison.loc[1:48]

comparison.plot(figsize=(12, 5))

plt.title("Weekday vs Weekend Electricity Demand (2025)")
plt.xlabel("Settlement Period")
plt.ylabel("Average National Demand (MW)")
plt.grid(True)
plt.show()
```

Conclusion 1: National demand for energy is lower on weekends

Does average electricity demand in Great Britain vary significantly between different months of the year?

```python
df["MONTH"] = df["SETTLEMENT_DATE"].dt.month
```

```python
df[["SETTLEMENT_DATE", "MONTH", "ND"]].head()
```

```python
monthly_demand = df.groupby("MONTH")["ND"].mean()
monthly_demand
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    monthly_demand.index,
    monthly_demand.values,
    marker="o"
)

plt.xticks(
    range(1, 13),
    ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
     "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
)

plt.xlabel("Month")
plt.ylabel("Average Electricity Demand (MW)")
plt.title("Average Monthly Electricity Demand in GB (2025)")

plt.grid(True)
plt.show()
```

```python
highest_month = monthly_demand.idxmax()
lowest_month = monthly_demand.idxmin()

highest_demand = monthly_demand.max()
lowest_demand = monthly_demand.min()

difference = highest_demand - lowest_demand

print("Highest month:", highest_month)
print("Lowest month:", lowest_month)
print("Difference:", difference, "MW")
```

```python
daily_profiles = df.groupby(
    ["MONTH", "SETTLEMENT_PERIOD"]
)["ND"].mean().unstack("MONTH")
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    daily_profiles.index,
    daily_profiles[1],
    label="January"
)

plt.plot(
    daily_profiles.index,
    daily_profiles[7],
    label="July"
)

plt.xlim(1, 48)

plt.xlabel("Settlement Period")
plt.ylabel("Average Demand (MW)")
plt.title("January vs July Daily Demand Profile (2025)")

plt.legend()
plt.grid(True)
plt.show()
```

Conlucsion on Jan vs July: 
Demand is higher in winter at every time of day. And this gap remains the smallest overnight and the largest during the daytime and evening. This point toward heating plus lighting as the main seasonal driver. In britain AC isn't as common so summer doesn't have a matching cooling load for ND.
January peak is higher and about 5,000 MW higher than daytime plateau and peaks around 17:30 - 18:00. Sunset which is around 16:30 lands at the same time people start getting home and using household electricity like lights, cooking and heating etc.
July peak is lower and flatter and it comes about an hour later. It's still light until around 21:00 so lighting adds little and peak is mainly cooking and domestic activities.
Winter has a bigger swing over the day:
Jan: Peak/trough = 37.1/22.2 = 1.67
July: Peak/trough = 28.9/19.3 = 1.50

So we can see that GB demand is strongly seasonal. Winter demand is roughly 15-30% higher depending on the time of day, and the evening peak is much sharper mainly because of heating and early darkness. Summer demand is lower and flatter and solar starts to reduce it around midday.
