```python
df["DEMAND_DIFFERENCE"] = df["TSD"] - df["ND"]

df["DEMAND_DIFFERENCE"].describe()
```

```python
#calculating average for ND by the SETTLEMENT_PERIOD
average_demand = df.groupby("SETTLEMENT_PERIOD")["ND"].mean()
average_demand.head(55)
```

```python
#focus on the 48 normal settlement periods
average_demand = average_demand.loc[1:48]

plt.figure(figsize=(12,5))

plt.plot(average_demand.index, average_demand.values)

plt.xlabel("Settlement Period")
plt.ylabel("Average Electricity Demand (MW)")
plt.title("Average Daily Electricity Demand in GB (2025)")

#SETTLEMENT PERIODS FOR CORRESPONDING TIME: 0 = 00:00, 10 = 05:00, 20 = 10:00, 30 = 15:00, 40 = 20:00, 48 = 00:00

plt.grid(True)
plt.show()
```

```python
#Comparing weekday and weekend electricity demand
df["SETTLEMENT_DATE"] = pd.to_datetime(
    df["SETTLEMENT_DATE"],
    format="mixed",
    dayfirst=True
)
```

```python
df["DAY_OF_WEEK"] = df["SETTLEMENT_DATE"].dt.dayofweek #extracting day of week as a number i.e monday = 0 ... sunday = 6.
df["DAY_TYPE"] = df["DAY_OF_WEEK"].apply(
    lambda day: "Weekday" if day < 5 else "Weekend"
)
```

```python
comparison = df.groupby(
    ["SETTLEMENT_PERIOD", "DAY_TYPE"]
)["ND"].mean() #getting mean ND for "Weekday" and for "Weekend"

comparison = comparison.unstack("DAY_TYPE")
```

```python
comparison = comparison.loc[1:48]

comparison.plot(figsize=(12, 5))

plt.title("Weekday vs Weekend Electricity Demand (2025)")
plt.xlabel("Settlement Period")
plt.ylabel("Average National Demand (MW)")
plt.grid(True)
plt.show()
```
